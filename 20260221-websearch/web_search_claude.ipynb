## AnthoropicのAPI

```python
import anthropic

client = anthropic.AsyncAnthropic()

async def web_search(search_query: str):
    prompt = "エージェントから呼ばれるWeb検索用のツールです。GoogleSearch toolを呼び出して、ユーザーのクエリに対してWeb検索を行い、検索結果を返してください。"
    message = await client.messages.create(
        model="claude-opus-4-6",
        max_tokens=1000,
        messages=[
            {
                "role": "assistant",
                "content": prompt,
            },
            {
                "role": "user",
                "content": search_query,
            }
        ],
        tools=[
            {
                "type": "web_search_20250305",
                "name": "web_search",
                "max_uses": 5,
                "user_location": {
                    "type": "approximate",
                    "city": "Tokyo",
                    "region": "Tokyo",
                    "country": "JP",
                    "timezone": "Asia/Tokyo",
                },
            }
        ]
    )
    return message.content
```

```python
await web_search("東京から金沢への行き方")
```

ちゃんと検索できている

## VertexAIを使うと

```python
import anthropic

client = anthropic.AsyncAnthropicVertex(project_id="nnyn-dev", region="global")

async def web_search(search_query: str):
    prompt = "エージェントから呼ばれるWeb検索用のツールです。GoogleSearch toolを呼び出して、ユーザーのクエリに対してWeb検索を行い、検索結果を返してください。"
    message = await client.messages.create(
        # モデルの指定が異なる
        # https://docs.cloud.google.com/vertex-ai/generative-ai/docs/partner-models/claude?hl=ja
        model="claude-sonnet-4-5@20250929",
        max_tokens=1000,
        messages=[
            {
                "role": "assistant",
                "content": prompt,
            },
            {
                "role": "user",
                "content": search_query,
            }
        ],
        tools=[
            {
                "type": "web_search_20250305",
                "name": "web_search",
                "max_uses": 5,
                "user_location": {
                    "type": "approximate",
                    "city": "Tokyo",
                    "region": "Tokyo",
                    "country": "JP",
                    "timezone": "Asia/Tokyo",
                },
            }
        ]
    )
    return message.content
```

```python
await web_search("東京から金沢への行き方")
```

VertexAIを使っても検索機能が機能している
