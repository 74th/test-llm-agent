```python
from google import genai
from google.genai import types

genai_client = genai.Client().aio

async def web_search(query: str):
    grounding_tool = types.Tool(
        google_search=types.GoogleSearch()
    )

    prompt = f"エージェントから呼ばれるWeb検索用のツールです。GoogleSearch toolを呼び出して、ユーザーのクエリに対してWeb検索を行い、検索結果を返してください。ユーザーのクエリ: {query}"

    config = types.GenerateContentConfig(
        tools=[grounding_tool],
        thinking_config=types.ThinkingConfig(thinking_budget=-1), # dynamic
        # max_output_tokens=1024,
    )

    response = await genai_client.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt,
        config=config,
    )

    return response.text
```

```python
await web_search("東京から金沢への行き方")
```

```python
await web_search("VS Code Meetupの最新イベント")
```
