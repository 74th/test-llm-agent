```python
from openai import OpenAI
client = OpenAI()

def web_search(
    query: str,
):
    prompt = f"AIエージェントから呼ばれるWeb検索をするための関数です。web_search toolを使って、以下のクエリに対してWeb検索を行い、検索結果を返してください。音声アシスタントのツールのため、URLは返答の必要がないです。1回で応答できるようにページの要約も含めてください。30秒程度で応答してください\nクエリ: {query}"
    response = client.responses.create(
        model="gpt-5-mini",
        tools=[{"type": "web_search"}],
        input=prompt,
    )

    return response.output_text
```

```python
web_search("東京から金沢への行き方")
```

```python
web_search('"VS Code Meetup" イベント')
```

```python
web_search('今日 2026年2月21日 さいたま市の天気')
```
