```python
import dotenv
import os
dotenv.load_dotenv()

LM_STUDIO_HOST = os.getenv("LM_STUDIO_HOST", "localhost")
LM_STUDIO_PORT = os.getenv("LM_STUDIO_PORT", "1234")
MODEL = "google/gemma-4-e4b"
```

```python
from agents import Agent, Runner, function_tool
from agents import OpenAIChatCompletionsModel, AsyncOpenAI, set_tracing_disabled

set_tracing_disabled(disabled=True)

client = AsyncOpenAI(
    base_url=f"http://{LM_STUDIO_HOST}:{LM_STUDIO_PORT}/v1",
    api_key="lm-studio",
)

model = OpenAIChatCompletionsModel(
    model=MODEL,
    openai_client=client,
)

@function_tool
def nod(strength: str = "normal") -> str:
    """頷く。strength は light, normal, strong。"""
    # ここで既存UIへWebSocket/HTTP/MQTT送信
    print(f"頷き: {strength}")
    return f"nod:{strength}"

@function_tool
def smile(intensity: str = "normal") -> str:
    print(f"笑顔: {intensity}")
    return f"smile:{intensity}"

@function_tool
def tilt_head(direction: str = "right") -> str:
    """首をかしげる。direction は left または right。"""
    print(f"首をかしげる: {direction}")
    return f"tilt_head:{direction}"

@function_tool
def calculate(expression: str) -> str:
    """安全な範囲の計算を行う。"""
    # 実装では asteval / simpleeval / sympy などを使う
    print(f"計算: {expression}")
    return "計算結果"

agent = Agent(
    name="communication_agent",
    instructions="""
あなたは音声会話用のコミュニケーションエージェントです。
自然な相づちや表情が必要なときだけ、nod, smile, tilt_head を使ってください。
身体表現toolは1ターンに最大10回まで。
調べ物や計算が必要なときだけ対応するtoolを使ってください。
最終的な発話は短めにしてください。
""",
    model=model,
    tools=[nod, smile, tilt_head, calculate],
)

result = await Runner.run(agent, "文鳥のかわいさについて話して")
print(result.final_output)
```
